# 1. Pushing Repos into GitHub using Git Bash

## Erasing bad local commits

```python
"""
Your local history is messy or ahead by multiple commits, and you want to wipe 
out all local commits since your last successful push
"""

git reset --mixed origin/main

"""
Force-aligns your local history to match exactly 
where GitHub (origin/main) currently is.
"""
```

## Create a new local repository

```python
# 1 Create a new single go-to directory in which you can store all the files, folders and sub-folders of a particular project
# 2 Create all the files and folders in that directory

# 4 Open Git bash and go to the project directory using the below command:
cd "<project path>"

# 3 Create an .gitignore file and write all the files you want to ignore tracking, committing and pushing into github repo
# (do not create the .gitignore file in .txt format)
touch .gitignore

# 5 Once you are inside the project directory inside the Git bash environment, check your current user name and user email
git config --list

# 6 If you are logged in as the appropriate user it is fine, else in any other case login using the below two lines of command:
git config user.name --global <your user name>
git config user.email --global <your user mailid>
# if you want to login to only one particular project, login as local
git config user.name --local <your user name>
git config user.email --local <your user mailid>

# Once logged-in, you can check your login credentials 
git config --global --list      # OR
git config --local --list      # OR    
    
# 7 Initialize git
git init

# 8 Check the status
git status

# 9 Add the project to the tracking state
git add .
git add --all
git add --A

# Once all the changes are done and you are satisfied, commit the branch
# (Committing is very important step before pushing the local repo to Github )    

# 9 Once again confirm the status of the repo:
git status

     
```

## Uploading(Pushing) the project(repo) to the Github

```python
"""
# 1.  Naming the main or master branch 
&
# 2.  Committing the changes

are very important steps before pushing the local repo to Github
"""     


"""
Step #1
"""
# Below command helps you remane your current local branch to 'main'
git branch -M <"your desired name for the branch">
# For example:
git branch -M main
"""
here is -M flag is for Moving/Renaming the branch
"""



"""
Step #2
"""
# After creating the new empty repo in the Github, copy the HTTPS link
git remote add origin <"New Repo's HTTPS link in Github">
# for an example:
git remote add origin https://github.com/USERNAME/MyProject.git



"""
Step #3
"""
# If there is any problem in creating the origin point, try the below command:
git remote set-url origin <"New Repo's HTTPS link in Github">



"""
Step #4
"""
# Check where origin is currently pointing:
git remote -v
# you'll see:
origin  <"New Repo's HTTPS link you've just entered above"> (fetch)
origin  <"New Repo's HTTPS link you've just entered above"> (push)



"""
Step #5
"""
# If the origin is pointing to the correct path, then your Step #2 succeeded, then proceed to Step #6.
# Else execute Step #3



"""
Step #6
"""
#Just before pushing a repo, check which all files are ignored from tracking and pushing
git status --ignored


"""
Step #7
"""
# Once you commit all the changes, the final step to upload the project to the Github repository
git push -u origin main
#OR
git push -u origin <"your desired name for the branch">
"""
Note: 

-u flag  is the shorthand notation for --set-upstream
This enables Git Bash to store the HTTPS link of the Github repo to where 
you are pushing the current local project to.

So you have to do this just Once and every time you make changes in the local repo, 
you just have to commit and push without the need for Github repo's HTTPS link everytime

git commit -m "committing in the local repo"
git push                                                       # to push the repo from the same local machine where you uploaded it for the first time
OR
git pull                                                       # if you want to pull your repo in the same local machine

""" 
```

## Pushing large files (> 100 MB)

```python
"""
Here we use Git LFS (Large File Storage) for pushing files above 100 MB to Github

For uploading the large files to Git LFS, first bring the large files to be uploaded to 'working directory' state 
whether they are in 'commit' or in 'staging area' state  
"""

"""
Step #0
"""
# If the files are in 'staging' area, in other words, if you are tracking the files, and to bring them back to 'working directory' or untrack them
git rm --cached <filename>      # to untrack a particular file
#OR
git rm --cached .               # to untrack all the files in the repo
#OR    
git rm --cached -r .            # to recursively untrack the files in the directories inside repo    

# If the files are already 'commited', you can undo that and bring them back to 'working directory' by below command
git reset HEAD~



"""
Step #1
"""
# First of all, download Git LFS (Git Large File Storage) from the below link
https://git-lfs.com/


"""
Step #2
"""
# Copy the large file in the required repo location, e.g., XYZ.mp4 


"""
Step #3
"""
# Go to the repo path in your local machine and install git lfs (only once per user account)
git lfs install


"""
Step #4
"""
# Track the large file
git lfs track "*.mp4"     # for tracking any generic .mp4 file


# Track the all the files in the repo
git add .

# To check whether Git LFS is tracking the required files  
git lfs status
    
# Commit the changes in the repo
git commit -m <"LFS commit message"> 


# Push     
git push -u origin <"name of the branch">  #OR
git push -u origin main  
#OR
git push origin <"name of the branch"> #OR
git push origin main  
```

# 2. Cloning and Pulling Repos from Github using Git Bash

##
Cloning (git clone) creates a new local copy of a repository in the Github, 
while Pulling (git pull) updates an existing local repository with changes from the remote repo in the Github


## Cloning

```python
# Cloning is done when you first create a repo on Github and bring it down to the local machine

# Once the repository is created on Github (along with .gitignore, readme.md and license, etc), copy the repo's https link

# In Git Bash, Go to the desired directory 
cd <"path of desired directory in the local machine where you want to store the repo">

# Clone the new repo
git clone <"HTTPS link of the new Github repo">
```

## Pulling

```python
# First open Git Bash where you want to pull the repository


# Initialize Git at that location
git init


git status
# check the status, you will be able to make out that the Git is initialized at that location


# Then connect to the Github repo (add the origin)
git remote add origin <"HTTPS link of the required Github repo">


# Now finally, you can pull the repository to the local system
git pull origin main
# OR
git pull origin <"The branch you want to pull">


```
