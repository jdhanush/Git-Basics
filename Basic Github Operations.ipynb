# 1. Pushing Repos into GitHub using Git Bash

## Create a new local repository

```python
# 1 Create a new single go-to directory in which you can store all the files and sub-folders of a particular project
# 2 Create all the files and folders in that directory

# 4 Open Git bash and go to the project directory using the below command:
cd "<project path>"

# 3 Create an .gitignore.txt file and write all the files you want to ignore tracking, committing and pushing into github repo
touch .gitignore.txt

# 5 Once you are inside the project directory inside the Git bash environment, check your current user name and user email
git config --list

# 6 If you are logged in as the appropriate user it is fine, else in any other case login using the below two lines of command:
git config user.name --global <your user name>
git config user.email --global <your user mailid>
# if you want to login to only one particular project, login as local
git config user.name --local <your user name>
git config user.email --local <your user mailid>

# Once logged-in, you can check your login credentials 
git config --global --list      # OR
git config --local --list      # OR    
    
# 7 Initialize git
git init

# 8 Check the status
git status

# 9 Add the project to the tracking state
git add .
git add --all
git add --A

# Once all the changes are done and you are satisfied, commit the branch
# (Committing is very important step before pushing the local repo to Github )    

# 9 Once again confirm the status of the repo:
git status

     
```

## Creating and viewing branches

```python
# To see all the branches:
git branch

# By default, you'll see a single 'main' branch, you can change the 'main' branch by:
git branch -M <alternate name to main>
```

## Uploading(Pushing) the project(repo) to the Github

```python
# After creating the new empty repo in the Github, copy the HTTPS link
git remote add origin <"New Repo's HTTPS link in Github">
# for an example:
git remote add origin https://github.com/USERNAME/MyProject.git


# If there is any problem in creating the origin point, try the below command:
git remote set-url origin <"New Repo's HTTPS link in Github">


# Check where origin is currently pointing:
git remote -v
# you'll see:
origin  <"New Repo's HTTPS link you've just entered above"> (fetch)
origin  <"New Repo's HTTPS link you've just entered above"> (push)


# If the origin is pointing the correct path, proceed to the next step, else redo the first two steps of this section



#Just before pushing a repo, check which all files are ignored from tracking and pushing
git status --ignored


# Once you commit all the changes, the final step to upload the project to the Github repository
git push -u origin main
#OR
git push -u origin <alternate name to main>
```
